```python
df = pd.read_csv('Dataset/financial_loan_data.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df['emp_title'].fillna('None', inplace=True)
```

```python
df['issue_date'] = pd.to_datetime(df['issue_date'])
```

```python
df['month_num'] = df['issue_date'].dt.month
df['month_name'] = df['issue_date'].dt.month_name()
```

```python
MTD_Month = 12
PMTD_Month = 11
```

```python
def format_currency(val):
    return f'Rs.{val:,.2f}'


def format_pct(val):
    return f"{val * 100:,.2f}%" if val <= 1.0 else f"{val:,.2f}%"
```

# Total Loan Applications

```python
Total_Loan_Applications = df['id'].count()
Total_MTD_Loan_Applications = df[df['month_num'] == MTD_Month]['id'].count()
Total_PMTD_Loan_Applications = df[df['month_num'] == PMTD_Month]['id'].count()
Total_Loan_Applications_MoM_Growth = round(((Total_MTD_Loan_Applications - Total_PMTD_Loan_Applications)/Total_PMTD_Loan_Applications)*100, 2)
```

```python
print(f'Total Loan Applications: {Total_Loan_Applications}')
print(f'Total MTD Loan Applications: {Total_MTD_Loan_Applications}')
print(f'Total PMTD Loan Applications: {Total_PMTD_Loan_Applications}')
print(f'Total Loan Applications MoM Growth: {Total_Loan_Applications_MoM_Growth}')
```

# Total Funded Amount

```python
Total_Funded_Amount = df['loan_amount'].sum()
Total_MTD_Funded_Amount = df[df['month_num'] == MTD_Month]['loan_amount'].sum()
Total_PMTD_Funded_Amount = df[df['month_num'] == PMTD_Month]['loan_amount'].sum()
Total_Funded_Amount_MoM_Growth = round(((Total_MTD_Funded_Amount - Total_PMTD_Funded_Amount)/Total_PMTD_Funded_Amount)*100, 2)
```

```python
print(f'Total Funded Amount: {format_currency(Total_Funded_Amount)}')
print(f'Total MTD Funded Amount: {format_currency(Total_MTD_Funded_Amount)}')
print(f'Total PMTD Funded Amount: {format_currency(Total_PMTD_Funded_Amount)}')
print(f'Total Funded Amount MoM Growth: {format_pct(Total_Funded_Amount_MoM_Growth)}')
```

# Total Amount Received

```python
Total_Amount_Received = df['total_payment'].sum()
Total_MTD_Amount_Received = df[df['month_num'] == MTD_Month]['total_payment'].sum()
Total_PMTD_Amount_Received = df[df['month_num'] == PMTD_Month]['total_payment'].sum()
Total_Amount_Received_MoM_Growth = round((((Total_MTD_Amount_Received - Total_PMTD_Amount_Received)/Total_PMTD_Amount_Received)*100), 2)
```

```python
print(f'Total Funded Applications: {format_currency(Total_Amount_Received)}')
print(f'Total MTD Funded Applications: {format_currency(Total_MTD_Amount_Received)}')
print(f'Total PMTD Funded Applications: {format_currency(Total_PMTD_Amount_Received)}')
print(f'Total Amount Received MoM Growth: {format_pct(Total_Amount_Received_MoM_Growth)} ')
```

# Average Interest Rate

```python
Average_Int_Rate = df['int_rate'].mean()
Average_MTD_Int_Rate = df[df['month_num'] == MTD_Month]['int_rate'].mean()
Average_PMTD_Int_Rate = df[df['month_num'] == PMTD_Month]['int_rate'].mean()
Average_Int_Rate_MoM_Growth = round(((Average_MTD_Int_Rate - Average_PMTD_Int_Rate)/Average_PMTD_Int_Rate)*100, 2)
```

```python
print(f'Total Interest Rate: {format_pct(Average_Int_Rate * 100)}')
print(f'Total MTD Interest Rate: {format_pct(Average_MTD_Int_Rate* 100)}')
print(f'Total PMTD Interest Rate: {format_pct(Average_PMTD_Int_Rate* 100)}')
print(f'Average Interest Rate MoM Growth: {format_pct(Average_Int_Rate_MoM_Growth)}')
```

# Average Debt to Interest

```python
Average_DTI = df['dti'].mean()
Average_MTD_DTI = df[df['month_num'] == MTD_Month]['dti'].mean()
Average_PMTD_DTI = df[df['month_num'] == PMTD_Month]['dti'].mean()
Average_DTI_MoM_Growth = round(((Average_MTD_Int_Rate - Average_PMTD_Int_Rate)/Average_PMTD_Int_Rate)*100, 2)
```

```python
print(f'Total Debt to Interest: {format_pct(Average_DTI * 100)}')
print(f'Total MTD Debt to Interest: {format_pct(Average_MTD_DTI* 100)}')
print(f'Total PMTD Debt to Interest: {format_pct(Average_PMTD_DTI* 100)}')
print(f'Average Debt to Interest MoM Growth: {format_pct(Average_DTI_MoM_Growth)}')
```

# Good Loan vs. Bad Loan

```python
Good_Loan = ['Fully Paid', 'Current']
Bad_Loan = ['Charged Off']
```

## Separating Good Loan and Bad Loan Data

```python
Good_Loan_df = df[df['loan_status'].isin(Good_Loan)]
Bad_Loan_df = df[df['loan_status'].isin(Bad_Loan)]
total_count = df['id'].count()
```

### Good Loan KPIs

```python
good_loan_pct = round((len(Good_Loan_df)/total_count)*100, 2)
good_loan_applications = Good_Loan_df['id'].count()
good_loan_funded_amount = Good_Loan_df['loan_amount'].sum()
good_loan_total_received_amount = Good_Loan_df['total_payment'].sum()
```

```python
print(f'Good Loan Applicants Percentage: {format_pct(good_loan_pct)}')
print(f'Total Good Loan Applications: {good_loan_applications}')
print(f'Total Good Loan Funded Amount: {good_loan_funded_amount}')
print(f'Good Loan Total Received Amount: {format_currency(good_loan_total_received_amount)}')
```

### Bad Loan KPIs

```python
bad_loan_pct = round((len(Bad_Loan_df)/total_count)*100, 2)
bad_loan_applications = Bad_Loan_df['id'].count()
bad_loan_funded_amount = Bad_Loan_df['loan_amount'].sum()
bad_loan_total_received_amount = Bad_Loan_df['total_payment'].sum()
```

```python
print(f'Bad Loan Applicants Percentage: {format_pct(bad_loan_pct)}')
print(f'Total Bad Loan Applications: {bad_loan_applications}')
print(f'Total Bad Loan Funded Amount: {bad_loan_funded_amount}')
print(f'Bad Loan Total Received Amount: {format_currency(bad_loan_total_received_amount)}')
```

### Loan Status Grid View

```python
loan_status_grid_view = df.groupby('loan_status').agg(
    Total_Loan_Applications = ('id', 'count'),
    Total_Funded_Amount = ('loan_amount', 'sum'),
    Total_Amount_Received = ('total_payment', 'sum'),
    Average_Interest_Rate = ('int_rate', 'mean'),
    Average_DTI = ('dti', 'mean')
).reset_index()
```

```python
columns = ['Total_Funded_Amount', 'Total_Amount_Received']
for col in columns:
    loan_status_grid_view[col] = loan_status_grid_view[col].apply(format_currency)
```

```python
pct_columns = ['Average_Interest_Rate', 'Average_DTI']
for p_col in pct_columns:
    loan_status_grid_view[p_col] = loan_status_grid_view[p_col].apply(format_pct)
```

# Function to Change Format for Analysis

```python
def change_format(dataframe, *args):
    for data in args:
        dataframe[data] = dataframe[data].apply(format_currency)
```

## Monthly Trends by Issue Date 

```python
monthly_detail = df.groupby('month_name').agg(
    Monthly_Total_Loan_Applications = ('id', 'count'),
    Monthly_Total_Funded_Amount = ('loan_amount', 'sum'),
    Monthly_Total_Amount_Received = ('total_payment', 'sum')
).reset_index().sort_values(ascending=False, by='Monthly_Total_Loan_Applications')

change_format(monthly_detail, 'Monthly_Total_Funded_Amount', 'Monthly_Total_Amount_Received')
monthly_detail
```

## Regional Analysis by State

```python
region_detail = df.groupby('address_state').agg(
    Regional_Total_Loan_Applications = ('id', 'count'),
    Regional_Total_Funded_Amount = ('loan_amount', 'sum'),
    Regional_Total_Amount_Received = ('total_payment', 'sum')
).reset_index().sort_values(ascending=False, by='Regional_Total_Loan_Applications')

change_format(region_detail, 'Regional_Total_Funded_Amount', 'Regional_Total_Amount_Received')
region_detail
```

## Loan Term Analysis 

```python
term_detail = df.groupby('term').agg(
    Term_Total_Loan_Applications = ('id', 'count'),
    Term_Total_Funded_Amount = ('loan_amount', 'sum'),
    Term_Total_Amount_Received = ('total_payment', 'sum')
).reset_index().sort_values(ascending=False, by='Term_Total_Loan_Applications')

change_format(term_detail, 'Term_Total_Funded_Amount', 'Term_Total_Amount_Received')
term_detail
```
