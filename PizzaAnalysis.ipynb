```python
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True)

df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['month_name'] = df['order_date'].dt.month_name()
```

```python
monthly_sales = (
    df.groupby(['year', 'month'])['total_price']
    .sum()
    .reset_index()
)

monthly_sales
```

```python
monthly_sales['MoM_Growth'] = (
    monthly_sales['total_price'].pct_change() * 100
)

monthly_sales
```

```python
monthly_sales['YoY_Growth'] = (
    monthly_sales['total_price'].pct_change(periods=12) * 100
)

monthly_sales
```

```python
latest_date = df['order_date'].max()

cmtd = df[
    (df['order_date'].dt.month == latest_date.month) &
    (df['order_date'].dt.year == latest_date.year)
]['total_price'].sum()

cmtd
```

```python
previous_month = latest_date.month - 1

pmtd = df[
    (df['order_date'].dt.month == previous_month) &
    (df['order_date'].dt.year == latest_date.year)
]['total_price'].sum()

pmtd
```

```python
cytd = df[
    df['order_date'].dt.year == latest_date.year
]['total_price'].sum()

cytd
```

```python
previous_year = latest_date.year - 1

pytd = df[
    df['order_date'].dt.year == previous_year
]['total_price'].sum()

pytd
```

```python
sales_by_day = (
    df.groupby(df['order_date'].dt.day_name())['total_price']
    .sum()
    .sort_values(ascending=False)
)

sales_by_day
```
