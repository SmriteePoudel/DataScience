```python
import matplotlib.pyplot as plt 
```

```python
''' 
X-axis -> Independent variable 
Y-axis -> Dependent variable
'''
```

LINE BAR

```python
products = ['MOBILE','BOTTLE','KEYBOARD','MOUSE']
sales=[400,250,300,270]
revenue=[300,200,150,50]
```

```python
plt.figure(figsize= (6,4)) #600*400 pixel
#linestyle-> solid, dashed, -- , dotted
plt.plot(products,sales, color= 'red', marker='o', linestyle='dashed', label ='sales')
plt.plot(products,revenue, color='blue', marker='*', linestyle='solid',label='revenue')
plt.xlabel('PRODUCTS')
plt.ylabel('SALES & REVENUE')
plt.title('Total Sales and Revenue of Product')
plt.legend()
plt.grid(linestyle='--')
plt.savefig('product.png')
plt.show() #ending point of visualization
```

```python
plt.figure(figsize= (6,4)) #600*400 pixel
#linestyle-> solid, dashed, -- , dotted
plt.plot(products,sales, color= 'red', marker='o', linestyle='dashed', label ='sales')
plt.plot(products,revenue, color='blue', marker='*', linestyle='solid',label='revenue')
plt.xlabel('PRODUCTS')
plt.ylabel('SALES & REVENUE')
plt.title('Total Sales and Revenue of Product')
plt.legend()
plt.grid(linestyle='--')
plt.savefig('product.png')
plt.close() #phoro gets saved but isnt visible in screen 
```

```python
for idx, value in enumerate('sales'):
    print(idx, value)
```

BAR GRAPH

```python

plt.figure(figsize= (6,5))
plt.bar(products,sales)

```

```python
plt.bar(products, sales,
        color=['red', 'green', 'blue', 'purple'],
        edgecolor='black',
        width=0.5)

plt.xlabel('PRODUCTS')
plt.ylabel('SALES & REVENUE')
plt.title('Total Sales and Revenue of Product')

for idx, val in enumerate(sales):
    # va -> top, bottom, center
    # ha -> left, right, center
    plt.text(products[idx], val, str(val),
             va='bottom', ha='center')

plt.show()
```

```python
import pandas as pd

df = pd.read_csv('Dataset/temperature.csv')
df.head()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Temperature (°C)'], df['Humidity (%)'],color ='green',edgecolor='black')

plt.xlabel('Temperature')
plt.ylabel('Humidity')
plt.title('Temperature vs Humidity')

plt.show()
plt.savefig('scatter.png')
```

```python
print(df.columns)
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Temperature (°C)'], df['Humidity (%)'])

plt.xlabel('Temperature (°C)')
plt.ylabel('Humidity (%)')
plt.title('Temperature vs Humidity')
plt.grid(True)

plt.show()
```

```python
import seaborn as sns

corr = df.corr(numeric_only=True)

sns.heatmap(corr, annot=True, cmap='coolwarm')

plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(df.index, df['Temperature (°C)'], marker='o', label='Temperature')
plt.plot(df.index, df['Humidity (%)'], marker='s', label='Humidity')

plt.xlabel('Observation')
plt.ylabel('Value')
plt.title('Temperature and Humidity Comparison')
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.boxplot(df['Temperature (°C)'])

plt.title('Temperature Box Plot')
plt.ylabel('Temperature (°C)')

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Temperature (°C)'], bins=10)

plt.title('Temperature Distribution')
plt.xlabel('Temperature (°C)')
plt.ylabel('Frequency')

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(df['Temperature (°C)'], label='Temperature')
plt.plot(df['Humidity (%)'], label='Humidity')

plt.title('Temperature and Humidity Trend')
plt.xlabel('Observation')
plt.ylabel('Value')
plt.legend()

plt.show()
```
