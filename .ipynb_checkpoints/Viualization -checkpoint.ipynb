```python
import matplotlib.pyplot as plt 
```

```python
''' 
X-axis -> Independent variable 
Y-axis -> Dependent variable
'''
```

LINE BAR

```python
products = ['MOBILE','BOTTLE','KEYBOARD','MOUSE']
sales=[400,250,300,270]
revenue=[300,200,150,50]
```

```python
plt.figure(figsize= (6,4)) #600*400 pixel
#linestyle-> solid, dashed, -- , dotted
plt.plot(products,sales, color= 'red', marker='o', linestyle='dashed', label ='sales')
plt.plot(products,revenue, color='blue', marker='*', linestyle='solid',label='revenue')
plt.xlabel('PRODUCTS')
plt.ylabel('SALES & REVENUE')
plt.title('Total Sales and Revenue of Product')
plt.legend()
```
