
'''1. Fill Data
   Numeric
   ------------
    a. Mean-> If data is in range fill with mean.
    b. Median-> If not in range fill with median 
-------------------
  Categorical
----------------------

   c. Mode -> Use 0th index while filling data
 --------------------

```python
df['COMMISSION_PCT'].median()
```

```python
df['COMMISSION_PCT'].mean()
```

```python
#Fillna
```

```python

''' Round by 2 '''
commission_mean= round (df['COMMISSION_PCT'].mean(),2)


```

```python
#inplace=False(default)
#One way to clean the data
#df['COMMISSION_PCT'].fillna(commission_mean, inplace = True)
```

```python
#Safe way to clean the data 

df['COMMISSION_PCT'] = df['COMMISSION_PCT'].fillna(commission_mean)

df
```
