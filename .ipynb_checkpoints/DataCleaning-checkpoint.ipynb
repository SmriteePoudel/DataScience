```python
df['COMMISSION_PCT'].median()
```

```python
df['COMMISSION_PCT'].mean()
```

```python
#Fillna
```

```python

''' Round by 2 '''
commission_mean= round (df['COMMISSION_PCT'].mean(),2)


```

```python
#inplace=False(default)
#One way to clean the data
#df['COMMISSION_PCT'].fillna(commission_mean, inplace = True)
```

```python
#Safe way to clean the data 

df['COMMISSION_PCT'] = df['COMMISSION_PCT'].fillna(commission_mean)

df
```

```python
'''DROP Data'''
df = pd.read_csv('Dataset/EMPLOYEES.csv')

```

```python
#dropna()
#Default -> row wise if any value null
#df.dropna(inplace =True) -> For permanent
#df.dropna() -> Temp
df = df.dropna() #-> Permanent
df 
```

```python
'''Parameters
------------------
axis -> 0 (default)-> Row and axis=1 -> Column
how = any(default)/all
inplace= False(default)/True
'''
```

```python


df = df.dropna(axis=1) 
df
```

```python
#Atleast one NULL drops the whole row/column
df = df.dropna(axis=1, how ='any') #Same as above
df
```

```python
#All value must be NaN to drop any row/column
df = df.dropna(axis=1, how ='all') 
df
```
