```python
import re
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns

#Classification Algorithm
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import confusion_matrix, classification_report
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer, ENGLISH_STOP_WORDS
```

```python
df = pd.read_csv('Dataset/movie_review.csv')
df.head()
```

```python
'''
Vectors
-----------
1. Count Vectorizer
2. Tfidf Vectorizer 
------------------------------
vectors = [
    unique words 
]

common_vector=[
    common words
]
unique_vectors=[
    unique words 
]
'''
```

```python
# ENGLISH_STOP_WORDS
```

```python
def cleanText(text):
    text = re.sub(r'<.*?>', ' ', text)
    text = re.sub(r'[^A-Za-z]', ' ',text)
    text = re.sub(r'\s+', ' ' ,text)
    return text
```

```python
df['review']=df['review'].apply(cleanText)
```

```python
df
```

```python
X= df['review']
Y=df['sentiment']
```

```python
X_train, X_test,Y_train, Y_test = train_test_split(
    X,Y, test_size=0.2, random_state= 42, stratify = Y
 

)
```

```python
stopwords={'do','no','not', 'i','better','beautiful'}
custom_stop_words=list(set(ENGLISH_STOP_WORDS-stopwords))
```

```python
tfidf = TfidfVectorizer(
    lowercase= True,
    stop_words=list(ENGLISH_STOP_WORDS),
    ngram_range=(1,2)
)  
X_train_tfidf=tfidf.fit_transform(X_train)
X_test_tfidf=tfidf.transform(X_test)
```

```python
model = MultinomialNB()
```

```python
model.fit(X_train_tfidf, Y_train)
```

```python
Y_pred= model.predict(X_test_tfidf)
```

```python
Y_pred
```

```python
cr = classification_report(Y_test, Y_pred)
cm = confusion_matrix(Y_test, Y_pred)
```

```python
print(cr)
```

```python
print(cm)
```

```python

sns.heatmap(
    cm,
    annot=True,
    fmt='.0f',
    xticklabels=['Predicted Healthy [0]', 'Predicted Unhealthy [1]'],
    yticklabels=['Actual Healthy [0]', 'Actual Unhealthy [1]']
)

plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.show()
```

```python
def predictReview(text):
    cleaned_text=cleanText(text)
    cleaned_tfidf= tfidf.transform([cleaned_text])
    prediction = model.predict(cleaned_tfidf)[0]
    return prediction
```

```python
predictReview('Movie is not good.')
```
