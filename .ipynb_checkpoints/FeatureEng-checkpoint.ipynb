```python
#Feature Engineering

import pandas as pd
```

```python
df = pd.read_csv('Dataset/HR_Dataset Refresh.csv')
df.head()
```

```python
df['Employee_Name'] = df['Employee_Name'].str.replace(',', ' ',regex=False)
df
```

```python
#To break column use split(find value,(n) no of column= index,expand=True)
```

```python
df[['last_name','first_name','middle_name']] = df['Employee_Name'].str.split(' ', n=2, expand = True)
df
```

```python
'''
#Fix it (How to make column in this pattern along side the names to be separated in the similar order. )
1. First Name 
2. Second Name
3. Middle Name manage the row ''' 
```

```python
df['PerformanceScore'].unique()
```

```python
ps ={
    'PIP': 0,
    'Needs Improvement': 1,
    'Fully Meets': 2,
    'Exceeds': 3
}
df['PerformanceScore_ID'] = df['PerformanceScore'].map(ps)
```

```python
df
```

```python
first_last=df['Employee_Name'].str.split(' ', n=1, expand=True)
first_last
```

```python
first_middle= first_last[1].str.strip().str.split(
    ' ', n=1, expand=True
)
first_middle
```

```python
df['first_name']=first_middle[0]
df['middle_name']=first_middle[1]
df['last_name']=first_last[0]
```

```python
df
```

```python
#Change in date

df['DateofHire'] = pd.to_datetime(df['DateofHire'], format='mixed')
```

```python
#Hired Year// Extracting Year
df['Hired_Year'] = df['DateofHire'].dt.year
```

```python
df
```

```python
#Extracting month
df['Hired_Month'] = df['DateofHire'].dt.month
```

```python
df
```

```python
df['Hired_Day'] = df['DateofHire'].dt.day
```

```python
df
```

```python
#Extracting name for month & day
df['Hired_Month'] = df['DateofHire'].dt.month_name()
df

```

```python
df['Hired_Day_name'] = df['DateofHire'].dt.day_name()
df
```
