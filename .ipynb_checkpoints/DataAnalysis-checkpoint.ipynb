```python
import pandas as pd
```

```python
df = pd.read_csv('Dataset/Employees.csv')
df.head()
```

```python
df['COMMISSION_PCT']=df['COMMISSION_PCT'].fillna(0.22)
df['MANAGER_ID']=df['MANAGER_ID'].fillna(100.0)
df['DEPARTMENT_ID']=df['DEPARTMENT_ID'].fillna(50.0)
```

```python
df['MANAGER_ID']=df['MANAGER_ID'].astype(int)
df['DEPARTMENT_ID']=df['DEPARTMENT_ID'].astype(int)
```

```python
df['HIRE_DATE']= pd.to_datetime(df['HIRE_DATE'])
```
