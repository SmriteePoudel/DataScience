```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
# Linear Regression Packages 
from sklearn.linear_model import LinearRegression 
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
df = pd.read_csv('Dataset/Real estate.csv')
df.head()
```

```python
cor = df.corr()
```

```python
plt.figure(figsize=(8,5))
sns.heatmap(cor, annot= True)
```

```python
# Feature must be multi-dimension
#Target must be one dimension
```

```python
df.columns
```

```python
feature = ['X2 house age',
       'X3 distance to the nearest MRT station',
       'X4 number of convenience stores', 'X5 latitude', 'X6 longitude']
target = 'Y house price of unit area'
```

```python
X= df[feature]
Y= df[target]
```

```python
#sns.boxplot(df[]
```

### Split Data

```python
'''
X -> Feature
Y-> Target

test _size -> split test size
random_state = 42 (index)
random_state -> data shuffle, 42th index extract data
'''
```

```python
X_train, X_test,Y_train, Y_test = train_test_split(
    X,Y, test_size=0.2, random_state= 42
)
```

```python
#(xi-mean)/ sd
scaler = StandardScaler()
```

```python
# fit_transform  -> feature training -
    #where, fit(learning) transform (Implement)
# transform (implementation) -> feature testing

X_train_scale = scaler.fit_transform(X_train)
X_test_scale = scaler.transform(X_test)
```

```python

X_train_scale
```

```python
X_train_scale
```

### Linear Regression Model

```python
model= LinearRegression()
```

```python
#Training model
# y = b0+b1.x1+ b2.x2+.......+bn.xn

model.fit( X_train_scale, Y_train)
```

```python
beta_0 = model.coef_
beta_1= model.intercept_
print(f'Slope/Coefficient:{beta_0}')
print(f'Intercept:{beta_0}')
```

```python
 Y_pred = model.predict( X_test_scale)
```

```python
 Y_pred
```

### Metrices

```python
mae = mean_absolute_error(Y_test, Y_pred)
mse = mean_squared_error(Y_test, Y_pred)
rmse = np.sqrt(mean_squared_error(Y_test, Y_pred))
r2 = r2_score(Y_test, Y_pred)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R² Score:", r2)         
```

### Visualization

```python
sns.regplot(x=Y_test, y=Y_pred, color='blue', line_kws={'color':'red'} )
```

```python
min_val = min(Y_test.min(),Y_pred.min())
max_val = min(Y_test.max(),Y_pred.max())
```

```python
plt.scatter(Y_test,Y_pred, color='blue')
plt.plot([min_val,max_val],[min_val,max_val], color='red')
```

```python
rf.fit(X_train_scale, Y_train)
rf_pred = rf.predict(X_test_scale)
```
