```python
import pandas as pd
```

```python
df = pd.read_csv('Dataset/Employees.csv')
df.head()
```

```python
df['COMMISSION_PCT']=df['COMMISSION_PCT'].fillna(0.22)
df['MANAGER_ID']=df['MANAGER_ID'].fillna(100.0)
df['DEPARTMENT_ID']=df['DEPARTMENT_ID'].fillna(50.0)
```

```python
df['MANAGER_ID']=df['MANAGER_ID'].astype(int)
df['DEPARTMENT_ID']=df['DEPARTMENT_ID'].astype(int)
```

```python
df['HIRE_DATE']= pd.to_datetime(df['HIRE_DATE'])
```

```python
emp_dep = (
    df.groupby('DEPARTMENT_ID')['EMPLOYEE_ID']
      .count()
      .reset_index()
      .rename(columns={
          'EMPLOYEE_ID': 'Count of employees',
          'DEPARTMENT_ID': 'Departments'
      })
      .sort_values(by='Count of employees', ascending=True,ignore_index=True)#Professionally used otherwise use reset= True
)
emp_dep
```

```python
job_emp = df.groupby('JOB_ID')['EMPLOYEE_ID'].count().reset_index().rename(columns= {'JOB_ID':'Job_Name','EMPLOYEE_ID':'Count of employee'})
    
```

```python
total_salary=df.groupby('DEPARTMENT_ID')['SALARY'].sum().reset_index().rename(columns = {'DEPARTMENT_ID':'Member of department','SALARY':'Salary'}).sort_values(ascending= False, by = 'Salary', ignore_index= True)
```

```python
total_salary.head()
```

```python
dep_job = (
    df.groupby(['DEPARTMENT_ID', 'JOB_ID'])['SALARY']
      .sum()
      .reset_index()
      .rename(columns={'DEPARTMENT_ID': 'dept', 'JOB_ID': 'job'})
)
dep_job
```

```python
dep_job_report = (
    df.groupby(['DEPARTMENT_ID', 'JOB_ID'])
      .agg(Total_Salary=('SALARY', 'sum'))
      .reset_index()
)

print(dep_job_report)
```

```python
dep_job_report['Total_Salary'] = dep_job_report['Total_Salary'].map("Rs.{:,.2f}".format)
dep_job_report
```

```python
columns = ['Max_Salary', 'Min_Salary', 'Avg Salary']

for col in columns:
    dep_job_report[col] = dep_job_report[col].map("Rs.{:,.2f}".format)
```

```python
dep_job_report.style.set_table_styles([
    {
        'selector': 'th',
        'props': [('text-align', 'center')]
    },
    {
        'selector': 'td',
        'props': [('text-align', 'center')]
    }
])
```

```python
weekdays=['Sunday','Monday','Tuesday','Wednesday','Thursday','Friday','Saturday']
df['HIRE_DAY']= pd.Categorical(df['HIRE_DAY'],categories = weekdays)
```

```python
emp_day = df.groupby(df['HIRE_DAY'], observed= False)['EMPLOYEE_ID'].count().reset_index()
emp_day
```
