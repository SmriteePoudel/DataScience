```python
columns= ['Position', 'State', 'Gender', 'MaritalDesc',
       'CitizenDesc', 'HispanicLatino', 'RaceDesc',
        'TermReason', 'EmploymentStatus', 'Department',
       'ManagerName', 'RecruitmentSource', 'PerformanceScore',
       'LastPerformanceReview_Date']
for col in columns:
    unique = df[col].unique()
    print(col)
    print('--'*35)
    print(unique)
    print("\n")
```

```python
#String Func
```

```python
#Strip() -> Used to remove unnecessary data like white spaces
df['Position']=df['Position'].str.strip()
```

```python
df['Gender']=df['Gender'].str.strip()
```

```python
#Removing all white spaces using for loop so that everything gets cleared at once.
columns= ['Position', 'State', 'Gender', 'MaritalDesc',
       'CitizenDesc', 'HispanicLatino', 'RaceDesc',
        'TermReason', 'EmploymentStatus', 'Department',
       'ManagerName', 'RecruitmentSource', 'PerformanceScore',
       'LastPerformanceReview_Date']
for s_col in columns:
    df[s_col]= df[s_col].str.strip()
```

```python
#To clean double yes no acceptance
df['HispanicLatino'].unique()
```

```python
#df['HispanicLatino']=df['HispanicLatino'].str.upper()
```

```python
h1 = {
    'no':'No',
    'yes':'Yes',
    'Yes':'Yes',
    'No':  'No'
    
}
df['HispanicLatino']=df['HispanicLatino'].map(h1)
```
