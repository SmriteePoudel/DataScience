```python
import re
```

```python
'''
re.sub(pattern, replace, value) -> to clean data or to remove pattern of words, digits, regex from data.
re.findall(pattern, value) -> to extract pattern of words, digits, regex from data.
'''
```

```python
text = "Hello world!!     abc@gmail.com, def123@gmail.com  . contact us: 9876543210 ###      or 01-5250453.       We are located in kathmandu."
```

```python
string_remove = re.sub(r'\w+', '', text)
string_remove
```

```python
number_remove = re.sub(r'\d', '', text)
number_remove
```

```python
ws_remove = re.sub(r'\s+', ' ', text)
ws_remove
```

```python
remove_has = re.sub(r'#', '', ws_remove)
remove_has
```

```python
remove_sc = re.sub(r'[#@!.,():]', '', ws_remove)
remove_sc
```

```python
remove_except_sc = re.sub(r'[^#@!.,():]', '', ws_remove)
remove_except_sc
```

```python
remove_email = re.sub(r'\w+\d*@\w+.\w+', '', text)
remove_email
```

```python
email = re.findall (r'\w+\d*@\w+.\w',text)
email
```

```python

tel_phone=re.findall(r'\d+-\d+',text)
tel_phone
```
