```python
from scipy import stats
```

Is the average total price of all pizza sales different from $15? 

```python
df=pd.read_csv('Dataset/pizza_sales.csv')
df.head()
```

```python
df.info()
```

```python
sample_mean=15
pop=df['total_price']
alpha= 0.05
```

```python
#stats.ttest_1samp(sample_mean, population)
stats.ttest_1samp(pop, popmean=sample_mean)
```

FORMULA

```python
#(actual_mean- sample_mean)/(sd)/root(total_number_of_data)
```

```python
#Formula & Working of the formula inside the function

import numpy as np
actual_mean=df['total_price'].mean()
sd=df['total_price'].std()
first_val= actual_mean - sample_mean
bot_val = sd/np.sqrt (48629)
```

```python
# stats.ttest_1samp(population, sample_mean)
t_stats, p_value=stats.ttest_1samp(pop, popmean=sample_mean)
```

```python
Ho='Ho:Average total price of all pizza sales is equal to $15'
Ha='Ha:Average total price of all pizza sales is not equal to $15'

print(f'{Ho}')
print(f'{Ha}')
print(f'Statistics:{t_stats:,.2f}')
print(f'Probability Value:{p_value:,.2f}')
print(f'Actual Average Price:Rs.{df['total_price'].mean():,.2f}')
print(f'Sample Average Price:{sample_mean}')
```

```python
if p_value<alpha:
    print(f"Decision failed to accept null hypothesis. \nConclusion: Needs to accept alternate hypothesis.\n{Ha}")
else:
    print(f"Decision failed to reject null hypothesis. \nConclusion: Needs to accept null  hypothesis.\n{Ho}")
```
