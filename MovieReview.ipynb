```python
def cleanText(text):
    text = re.sub(r'<.*?>', '', text)
    text = re.sub(r'[^A-Za-z]', '',text)
    text = re.sub(r'\s+', ' ' ,text)
    return text
```

```python
df['review']=df['review'].apply(cleanText)
```
